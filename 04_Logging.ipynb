## Python Logging

`Logging` is an important technique used to record information about what is happening inside a program.

Instead of using `print()` everywhere to understand what your program is doing, Python provides the built-in logging module.

```python
import logging
```

## 1. What is Logging?

`Logging` means recording messages about the execution of a program.

These messages can tell us:

- What the program is doing
- Whether an operation was successful
- Whether something unexpected happened
- Where an error occurred
- Important events that happened during execution

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.info("Application started")
logging.warning("Low memory")
logging.debug("Hello")
logging.error("Unable to connect to database")
```

## Why Use Logging?

`Logging` is especially useful when developing and maintaining real-world applications.

Without logging

We might use:

```python
print("User logged in")
print("Database connected")
print("Error occurred")
```

With logging:

```python
logging.info("User logged in")
logging.info("Database connected")
logging.error("Database connection failed")
```

### Logging provides additional features such as:

- Different severity levels
- Timestamps
- Writing logs to files
- Formatting messages
- Debugging production applications
- Recording errors
- Log rotation

## 3. print() vs logging

Both can display information, but they serve different purposes.

Using `print()`

```python
print("Application started")
print("User logged in")
print("Database error")
```

Using `logging`

```python
import logging
logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.info("User logged in")
logging.error("Database error")
```

| `print()`                         | `logging`                                |
| --------------------------------- | ---------------------------------------- |
| Mainly for simple console output  | Designed for application event recording |
| No log levels                     | Has multiple log levels                  |
| No built-in timestamps            | Can include timestamps                   |
| Mainly console output             | Console, files, and other handlers       |
| Difficult to control by severity  | Can filter by severity                   |
| Useful for quick debugging/output | Better for real applications             |


## 4. Log Levels

Log levels indicate the importance or severity of a message.

Python's standard logging levels include:

DEBUG --> INFO --> WARNING --> ERROR --> CRITICAL


## 5. DEBUG

`DEBUG` is used for detailed information useful during development and troubleshooting.


```python
import logging 

logging.basicConfig(level=logging.DEBUG)

logging.debug("Starting database connection")
logging.debug("Checking user credentials")
```

## 6. INFO

`INFO` indicates that something normal and expected happened.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.info("User logged in")
logging.info("Database connected")
```

## 7. WARNING

`WARNING` indicates that something unexpected or potentially problematic happened, but the program can continue.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.warning("Disk space is running low")
```

## 8. ERROR

`ERROR` indicates that an operation failed.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

try:
    result = 10 / 0

except ZeroDivisionError:
    logging.error("Division by zero occurred")
```

## 9. CRITICAL

`CRITICAL` represents a very serious error that may prevent the application from continuing normally.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.critical("Database server is completely unavailable")
```

## 10. Understanding Log Levels Together

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debugging information")
logging.info("Application started")
logging.warning("Memory usage is high")
logging.error("File could not be opened")
logging.critical("System failure")
```

## 11. Log Formatters

A `formatter` controls how a log message looks.

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s",
 
)

logging.debug("This is a debug message")
logging.info("Application started")
logging.warning("Low memory")
logging.error("Something went wrong")
logging.critical("Critical error")
```

| Format          | Meaning       |
| --------------- | ------------- |
| `%(asctime)s`   | Date and time |
| `%(levelname)s` | Log level     |
| `%(message)s`   | Log message   |
| `%(name)s`      | Logger name   |
| `%(filename)s`  | File name     |
| `%(lineno)d`    | Line number   |


## 12. File Handlers

A `handler` determines where log messages should go.

A `FileHandler` swrites logs to a file.

```python
import logging

logger = logging.getLogger("my_app")

logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("Application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)

logger.addHandler(file_handler)

logger.info("Application started")
logger.error("Something went wrong")
```

## 13. Console Handlers

A StreamHandler can send logs to the console.

```python
import logging

logger = logging.getLogger("my_app")

logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)

logger.addHandler(console_handler)

logger.info("Application started")
logger.error("Something went wrong")
```

## 15. Rotating Logs — Bonus

Log files can become very large if an application runs for a long time.

`Rotating logs` automatically creates new log files when the existing file reaches a specified size.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("application")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "application.log",
    maxBytes=1000,
    backupCount=3
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)

logger.addHandler(handler)

for i in range(100):
    logger.info("This is log message %s", i)
```
