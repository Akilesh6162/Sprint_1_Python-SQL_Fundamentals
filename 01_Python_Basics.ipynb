## Basic Concepts

### Variable

A **variable** is a named container used to store or refer to a value in a program. It allows us to save information and use that information later.

In Python, we do not need to declare the variable type separately. Python identifies the data type from the value assigned to the variable.

### Real-world analogy

Think of a variable like a **labelled box**.

- A box labelled `name` can contain `"Akilesh"`.
- A box labelled `age` can contain `23`.
- If the value changes, we can put a new value into the same labelled box.





```python
name = "Akilesh"
age = 23
city = "Chennai"
```

- `name`, `age`, and `city` are variables.
- Each variable stores a different value.
  
Variables are useful because values can be reused and changed while a program is running.

### Data Types

A **Data Type** tells Python what kind of data is contained in a variable. Each variable contains a different type of data.

In Python, data type is set when you assign a value to a variable. data type tells whether the values are `Text`, `Whole number`, `Decimal number` and `True/False`.

| Data Type | Example | Purpose |
|---|---|---|
| `int` | `25` | Whole numbers |
| `float` | `25.5` | Decimal numbers |
| `str` | `"Python"` | Text |
| `bool` | `True` | True/False values |
| `list` | `[10, 20, 30]` | Ordered, changeable collection |
| `tuple` | `(10, 20, 30)` | Ordered, generally immutable collection |
| `set` | `{10, 20, 30}` | Collection of unique values |
| `dict` | `{"name": "Alex"}` | Key-value pairs |

### type()
In Python, to check the data type of a variable using the `type()` function.

#### Example


```python
num = 30
print(type(num))
```

Python is an object-orientated language, so it uses classes to define data types.

### Operators

**Operators** are symbols or keywords used to perform operations on values.

Python divides the operators in the following groups:

#### 1. Arithmetic operators
Used for calculations:
`+`, `-`, `*`, `/`, `//`, `%`, `**`
##### Example:

```python
a=10
b=5
print(a*b)
```

#### 2. Comparison operators
Used to compare values:
`==`, `!=`, `>`, `<`, `>=`, `<=`

They return a Boolean result such as `True` or `False`.
##### Example:

```python
print(a==b)
print(a!=b)
```

#### 3. Logical operators
Used to combine conditions:
`and`, `or`, `not`
##### Example:

```python
print(a < 5 and  b < 10)
print(a < 5 or  b < 10)
print(not(a < 5 and  b < 10))
```

#### 4. Assignment operators
Used to assign or update values:
`=`, `+=`, `-=`, `*=`, `/=`
##### Example:

```python
x = 5

x += 10

print(x)
```

### 5. Membership operators
Used to check whether a sequence is presented in an object:
`in`, `not in`
##### Example:

```python
domain = ["Python Developer","AI/ML engineer","Java Developer","Front end",".Net"]
print("AI/ML engineer" in domain)
```

### 6. Identity operators
Used to check whether two variables refer to the same object, with the same memory location:
`is`, `is not`
##### Example:

```python
Technologies = domain
domain1 = ["Python Developer","AI/ML engineer","Java Developer","Front end",".Net"]
print(domain is Technologies)
print (domain is domain1)
```

*Operators allow a program to calculate values, compare information, make decisions, and manipulate data.*

### Comments
**Comments** are notes written inside source code for humans to understand the program. Python does not execute normal comments as program instructions. It can be used to temporarily prevent code from executing during testing.

Comments start with a `#` which is a single-line comment, and Python will ignore them.


```python
#This Line is Commented
print("Hello Python!!!")
```

Python does not have multi-line comments. It will use `#` for each line.

```python
#This is a comment
#written in
#more than just one line
print("Multiline comments!")
```

Alternatively, we can use a multiline string. Since Python ignores unassigned string literals, multiline strings can be used to add longer comments to the code.

```python
"""This is a comment
written in
more than just one line"""
print("Multiline Comments Using String Literals!")
```

## Control Flow

### IF/ELSE/ELIF

if, elif, and else help Python make decisions.

- if – Executes a block of code when a condition is true.
- elif – Checks another condition if the previous if or elif condition is false.
- else – Executes a block of code when none of the conditions are true.

Real-time analogy:

- If it is raining, take an umbrella.

- Else if it is sunny, wear sunglasses.

- Else – carry on normally.

Python checks each condition one by one. As soon as one condition is true, it performs that action.
##### Example:

```python
age = int(input("Enter Your Age:"))

if age < 18:
    print("Minor")
elif age == 18:
    print("Just became an Adult")
else:
    print("Adult")
```

### For Loop

A **for loop** is used to iterate over a sequence, i.e., a list, a tuple, a set, a dict, and a string.

real-time analogy:

- Imagine a teacher checking the attendance of every student in a classroom.

- The teacher has a list of students and checks them one by one until everyone has been checked.

```python
students = ["Akilesh", "Aadharsh", "Ajay R", "Chetan", "Chandini"]

for student in students:
    print(student,"is present")
```

### While Loop
A **while loop** is used to repeatedly execute a block of code as long as a given condition is `true`.

real-time analogy:
A while loop is like continuing to add water to a tank while the tank is not full and stopping once the tank becomes full.


```python
secrete_number = 62

while True:
    user_input = int(input("Enter Your Guess:"))
    if secrete_number == user_input:
        print("Yes, Your Guess is Right")
        break
    else:
        print("Sorry, Your Guess is Wrong")
```

This code is a simple number guessing game.

- Stores the secret number 62 in the variable `secret_number`.

- `while True` - So, the loop will continue running indefinitely until we explicitly stop it using break.

- `input()` asks the user to enter a number.
`int()` converts the user's input from a string into an integer.
The entered number is stored in `user_input`.

- Checks whether the user's guess is equal to the secret number.
If the guess is 62, this message is displayed.
Stops the while loop because the user guessed correctly.
If the guess is not equal to 62, the else block runs.
Displays a message saying the guess is incorrect. Then the loop starts again and asks for another guess.

The program keeps asking the user to guess the secret number until the user enters 62. Once the correct number is entered, break stops the loop.
