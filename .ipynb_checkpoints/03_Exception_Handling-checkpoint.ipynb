**Step 1- Understand the Concept**

Own words—`Exeception handling` is used to handle errors in our program.

Exception handling prevents our program from crashing by giving meaningful messages to users and allowing the program to continue running.

Errors are two types—

1. `Syntax Error` - Mistakes in writing the program, such as missing : or incorrect spelling. It happens before execution.

2. `Runtime Error` - The syntax is correct, but an error happens while the program is running.

**Common Built-in Exceptions**

|Exception         |           Description|
|---|---|
|ZeroDivisionError      |       Raised when dividing by zero|
|ValueError             |      Raised when a function gets a valid type but an invalid value|
|TypeError              |      Raised when an operation is performed on incompatible data types|
|NameError              |      Raised when a variable is not defined|
|IndexError             |     Raised when a list index is out of range|
|KeyError               |     Raised when a dictionary key is not found|
|AttributeError         |     Raised when an object does not have the requested attribute|
|FileNotFoundError      |      Raised when a file does not exist|
|ImportError            |    Raised when something cannot be imported from a module|
|ModuleNotFoundError    |      Raised when a module is not found|

**Step 2 – Demonstrate the Concept**

- `try` - We put the code which may cause an error inside try.

- `except` - If an error occurs, except handles the error and gives a meaningful message.

- `else` - else runs when there is no error.

- `finally` - finally runs whether an error occurs or not.

- `raise` - raise is used to create an exception manually.

- `assert` - assert is used to check whether a condition is true.

- `Custom Exceptions` - We can create our own exception class according to our requirement.

```python
# using try and except

try:
  print(x)
except:
  print("An exception occurred")
```

**Step 3 – Using try, except, else and finally**

`try:`
The code which may produce an error is written inside try.

`except:`
It handles the error when an exception occurs.

`else:`
It executes when there is no exception.

`finally:`
It executes whether an exception occurs or not.

```python
# using try, except, else and finally

try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Please enter a valid number")

except ZeroDivisionError:
    print("Number cannot be zero")

else:
    print("Result:", result)

finally:
    print("Execution completed")
```

**Step 4 – Raise**

`"Raise"` is used when we want to create an exception manually.

For example, if the user enters zero as the second number, we can manually raise ZeroDivisionError.

```python
# using raise

balance = 5000

withdraw = int(input("Enter a withdraw amount:"))
try:
    if withdraw > balance:
        raise ValueError("Insufficient balance")

except ValueError as e:
    print("Error:",e)
else:
    print("Amount successfully withdraw.")

finally:
    print("Program completed.")
```

**Step 5 – Assert**

`assert` is used to check whether a condition is true.

If the condition is true, the program continues.

If the condition is false, Python raises AssertionError.

Example:

assert balance >=0, "Balance cannot be Negative"

```python
# using assert

try:
    balance = int(input("Balance:"))

    assert balance >=0, "Balance cannot be Negative"

except AssertionError as e:
    print("Assertion error:",e)
else:
    print("Balance amount is",balance)
finally:
    print("Process completed.")
```

**Step 6 – Custom Exceptions**

`Custom Exceptions` are user-defined exceptions.

We create our own exception class by inheriting from Exception.

For example:

PasswordError

We can use PasswordError when the password does not satisfy our requirement.

```python
# Custom Exception

class InsufficientBalanceError(Exception):
    pass

balance = 10000
withdraw = int(input("Enter a withdrawal amount:"))

try:
    if balance < withdraw:
        raise InsufficientBalanceError("Insufficient balance")
except InsufficientBalanceError as e:
    print("Error:",e)
else:
    balance -= withdraw
    print("Balance is",balance)

finally:
    print("Transaction Completed.")
```

### ZeroDivisionError

**When it occurs:**

`ZeroDivisionError` occurs when we try to divide a number by zero.

Example:

10 / 0

**Correct way to handle:**

Use try and except ZeroDivisionError.

```python
# ZeroDivisionError

try:
    a = (int(input("Enter a number")))
    b = (int(input("Enter a number")))
    div = a/b
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Divided Value",div)
finally:
    print("Program Completed")
```

### ValueError

**When it occurs:**

`ValueError` occurs when a function receives the correct type of value but an inappropriate value.

Example:

int("abc")

The string "abc" cannot be converted into an integer.

**Correct way to handle:**

Use try and except ValueError.

```python
# ValueError

try:
    a = int(input("Enter a num:"))
    square = a * a

except ValueError as e:
    print("Error:",e)
else:
    print("Square :",square)
finally:
    print("Completed")
```

### TypeError

**When it occurs:**

`TypeError` occurs when an operation is performed on an inappropriate data type.

Example:

"10" + 5

A string and an integer cannot be added directly.

**Correct way to handle:**

Use try and except TypeError.

```python
# TypeError
try:
    number1 = 10
    number2 = input("Enter a number:")
    result = number1 + number2
except TypeError as e:
    print("Error:",e)
else:
    print("result:",result)
```

### NameError

**When it occurs:**

`NameError` occurs when we try to use a variable that has not been defined.

Example:

print(name)

if name was never created, Python gives NameError.

**Correct way to handle:**

Check the variable name and use try and except NameError when required.

```python
# NameError

try:
    print(username)
except:
    print("Variable name is incorrect")
```

### IndexError

**When it occurs:**

`IndexError` occurs when you try to access a list or sequence using an index that does not exist.

Example:

fruits = ["Apple", "Banana", "Mango"]

fruits[5]

The list has indexes from 0 to 2.

**Correct way to handle:**

Use try and except IndexError.

```python
# IndexError

try:
    collection = ["akilesh",34,44,True,"Vijay"]
    print(collection[9])
except:
    print("Index error")
```

### KeyError

**When it occurs:**

KeyError occurs when we try to access a dictionary using a key that does not exist.

Example:

marks = {"Math": 90, "Science": 85}

marks["Social"]

The key "Social" does not exist.

**Correct way to handle:**

Use try and except KeyError.

```python
        # KeyError

try:
    person={
        "name" : "Akilesh",
        "age"  : 23,
        "Fav_movie" : "Ghilli"
    }
    print(person["salary"])
except KeyError as e:
    print("Subject not found.",e)
```

### AttributeError

**When it occurs:**

AttributeError occurs when an object does not have the attribute or method that we are trying to use.

Example:

num = 100

num.upper()

The upper() method is not available for an integer.

**Correct way to handle:**

Use try and except AttributeError.

```python
# AttributeError

try:
    name = "John"
    name.append(" Smith")

except AttributeError:
    print("This object does not have that method")
```

### FileNotFoundError

**When it occurs:**

FileNotFoundError occurs when we try to open a file that does not exist.

Example:

open("students.txt", "r")

If students.txt does not exist, Python gives FileNotFoundError.

**Correct way to handle:**

Use try and except FileNotFoundError.

```python
# FileNotFoundError

try:
    file = open("employee.txt", "r")
    content = file.read()
    print(content)

except FileNotFoundError:
    print("File was not found")
```

### ImportError

**When it occurs:**

`ImportError` occurs when Python cannot import a requested name from a module.

Example:

from math import square

The math module does not have a function called square.

**Correct way to handle:**

Use try and except ImportError.

```python
# ImportError

try:
    from math import hello

except ImportError:
    print("Could not import the requested item")
```

### ModuleNotFoundError

**When it occurs:**

`ModuleNotFoundError` occurs when we try to import a module that is not installed or does not exist.

Example:

import mymodule

If mymodule is not available, Python gives ModuleNotFoundError.

**Correct way to handle:**

Use try and except ModuleNotFoundError.

```python
# ModuleNotFoundError

try:
    import my_unknown_module

except ModuleNotFoundError:
    print("Module does not exist.")
```

**Final Understanding**

- try - Contains the code which may cause an exception.

- except - Handles the exception.

- else - Runs when there is no exception.

- finally - Runs whether an exception occurs or not.

- raise - Used to manually create an exception.

- assert - Used to check whether a condition is true.

- Custom Exception - Used to create our own exception according to our requirement.
