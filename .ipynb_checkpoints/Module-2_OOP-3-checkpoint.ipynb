### Object-Oriented Programming (OOP) in Python

Object-Oriented Programming is a programming approach where we organize code around objects and classes.

For example, in a company application, we might have:

- Employee
- Manager
- Department
- Company

Each object can have data (attributes) and behavior (methods).

## 1. Class

### What is a Class?
A **class** is a blueprint/template used to create objects.

### Real-world analogy
think of a building plan.

an architect creates a blueprint that says:

* how many rooms?

* where are the doors?

* where are the windows?

the blueprint itself isn't a building. it is a plan for creating buildings.

similarly, a Python class is a blueprint for **creating objects.**

## concept
A class defines:

* what data an object will have
* what actions an object can perform


### Why use it?
Classes group related **data (attributes)** and **behavior (methods)** together.

### Syntax
```python
class Student:
    pass
```

````python
class student:
    name = "akilesh"
    std = "12th"
student1 = student()
print("Name of the student",student1.name)
print("Class:",student1.std)
````

**Key idea:** The class describes what a Student object can contain or do.


## 2. Object

An **object** is an instance of a class.

## Real-world analogy

The architect's blueprint can be used to construct many actual buildings.

similarly, one class can create many objects.

             Student Class
                   ↓
        ┌──────────┼──────────┐
        ↓          ↓          ↓
     student1   student2   student3
      Aadharsh      vijay      gowtham
### Why use objects?
A class is the blueprint; objects are the actual entities created from that blueprint.


````python
class car: # create a Class
    car_model = "Hyundai"
    car_color = "Green"
    car_milage = "20 km"

    def move_forwar(self):
        print("Moving forward!!")
    def move_backward(self):
        print("Moving Backward!!!")

car1 = car() # Creating Object 1
car2 = car() # Creating Object 2
car3 = car() # Creating Object 3

print(car1.car_model)
print(car2.car_milage)
print(car3.car_color)
````

## 3. Constructor – `__init__()`

`__init__()` is a special method that runs automatically when an object is created.

## Real-world analogy
imagine you're creating a employee ID card.

When a new students joins, you immediately need to enter:

* Name
* Age
* Department
  
The initial information is set when the student is created.
### Why use it?
It initializes an object's starting data.


````python
class car:

    def __init__(self,car_model,car_milage,car_color,car_manufacture_year):
        self.car_model = car_model
        self.car_milage = car_milage
        self.car_color = car_color
        self.car_manufacture_year = car_manufacture_year

    def move_forwar(self):
        print("Moving forward!!")
    def move_backward(self):
        print("Moving Backward!!!")
        

car1 = car("Hyundai",25,"grey",2024) # Creating Object 1
car2 = car("Honda City",22,"green",2025) # Creating Object 2
car3 = car("Thor",20,"black",2020) # Creating Object 3

print("Name of the car 1:",car1.car_model)
print("Milage of the car 1:",car1.car_milage)
print("Name of the car 2:",car2.car_model)
print("Milage of the car 2:",car1.car_milage)
print("Name of the car 3:",car3.car_model)
print("Year car3:",car3.car_manufacture_year)
````

`self` refers to the **current object**.


## 4. Instance Variables

Instance variables belong to a **particular object**. Different objects can have different values.

## Real-world analogy
Consider three students:

* Akilesh → Age 21
* vijay → Age 22
* gowtham → Age 20

Each students has their own age.

so age belongs to a particular students.

## Concept

Instance variables are variables that belong to a **specific object.**



````python
class car:

    def __init__(self,car_model,car_manufacture_year):
        self.car_model = car_model
        self.car_manufacture_year = car_manufacture_year

    def move_forwar(self):
        print("Moving forward!!")
    def move_backward(self):
        print("Moving Backward!!!")
car1=car("Hyundai",2025)
car2=car("BMW",2026)

print("Name:",car1.car_model,"| Year:",car1.car_manufacture_year)
print("Name:",car2.car_model,"| Year:",car2.car_manufacture_year)
````

## 5. Class Variables

A class variable is shared by all objects of  class.

## Real-world analogy

Suppose all employees belong to the same company:

* Akilesh  → stackly
* vijay  → stackly
* gowtham  → stckly

The company name is common to everyone.
 instead of storing the same information separately for every object, we can create a *class variable*
 
### When to use it?
it can be used when a value should be common to every instance.


````python
class car:
    car_name="Hyundai"

    def __init__(self,model):
        self.model = model

car1 = car("Verna")
car2 = car("Exter")
car3 = car("i20")

print("Company Name:",car1.car_name ,"model:",car1.model)
print("Company Name:",car2.car_name ,"model:",car2.model)
print("Company Name:",car3.car_name ,"model:",car3.model)
````

## 6. Instance Methods

Instance methods operate on object-specific data and normally receive `self` as the first parameter.

## Real-world analogy

A employee can perform actions:

* prepare
* attend meeting
* completing work

  These actions belongs to the student. 

````python
class student:

    def __init__(self,name,mark):
        self.name= name
        self.mark= mark

    def score(self):
        print(self.name,"scored",self.mark)

student1 = student("Akilesh",99)
student2 = student("Joseph Vijay", 100)

student1.score()
student2.score()
````

## 7. Static Methods

A static method does not need `self` or `cls`. It does not depend on an object's data or the class's data.

Use `@staticmethod`.

## Real-world analogy

Suppose a college has a utility function:

Check whether a number is even.

This doesn’t depend on any particular student.

It doesn’t need student information.

That’s a good use case for a static method.

**When to use?**

Use a static method when a function logically belongs to a class but doesn't need access to the object or class.


````python
class Calculator:
    @staticmethod
    def sq(a):
        return a*a

print(Calculator.sq(10))
````

## 8. Class Methods

A class method receives the class itself as `cls`.

Use `@classmethod`.

### When to use it?
Common uses include modifying class-level data and creating alternative constructors.

## Real-world analogy

Imagine a college changes its name.

The change applies to the college as a whole, not just one student.

That’s class-level behavior.


````python
class Student:

    school = "ABC School"

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school


Student.change_school("XYZ School")

print(Student.school)
````

## 9. Properties – `@property`

A property allows a method to be accessed like an attribute.

It is especially useful when you want controlled access to internal data.

## Real-world analogy

Think about a bank account.

You may want to allow someone to see the balance, but you don’t want them to directly modify it however they want.

A property allows controlled access.


````python
class Employee:

    def __init__(self, salary):
        self._salary = salary

    @property
    def salary(self):
        return self._salary


employee = Employee(50000)

print(employee.salary)
````

## 10. Encapsulation

**Encapsulation:** Binding data and methods together into a single unit (class). 
Also called: “Data Hiding”

Python uses naming conventions such as:
- `name` → public
- `_name` → protected-by-convention
- `__name` → name-mangled/private-by-convention
### Real-Life Example 
ATM machine: 
- You can withdraw money 
- But internal bank processing is hidden 

Benefits 
- Security 
- Controlled access 
- Easy maintenance

that's the basic idea of **encapsulation.**

**Key point:** *“Encapsulation protects data from direct unauthorized access.”*


````python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)

account.deposit(500)

print(account.get_balance())
````

## 11. Abstraction

**Abstraction** hides implementation details and exposes only the essential interface.

Python provides the `abc` module for formal abstract classes.

## Real-world analogy.
when you drive a car, you press:
    
     START
     
You don't need to understand:
   * Fuel injection
   *    ↓
   * Battery
   *    ↓
   * Starter motor
   *    ↓
   * Engine combustion
   *    ↓
   * Transmission


The complicated implementation is hidden.

Abstraction= show what is needed, hide how it works internally.
     

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car starts with a key or button.")

car = Car()
car.start()
````

### Output
```text
Car starts with a key or button.
```

An abstract method defines **what** a subclass must provide without defining the complete implementation.
