## 1. Database Design

### 1.1 What is a Database?

A `database` is an organized collection of data that allows us to store, manage, retrieve, and update information efficiently.

#### Real-time example

A college may need to store:

- Course details
- Teacher details
- Exam marks
- Attendance

Instead of maintaining this information in separate Excel files, we can store it in a database.

### 5. Relationships

A `relationship` in SQL describes how two or more tables are connected to each other.

For example, in a college database, we may have:

- `Students table`
- `Courses table`

A student can enroll in a course, so there is a relationship between these tables.

### 1. One-to-One (1:1)

One record in Table A is related to only one record in Table B.

Example: One person has one passport.

*`UNIQUE` ensures that one person cannot have multiple passport records in this relationship.*

### 2. One-to-Many (1:N)

One record in Table A can be related to many records in Table B.

This is the most common relationship.

Example: One department can have many employees.

Example data:

**Department**

| department_id | department_name |
| ------------- | --------------- |
| 1             | IT              |
| 2             | HR              |


**Employee**

| employee_id | employee_name | department_id |
| ----------- | ------------- | ------------- |
| 101         | Arun          | 1             |
| 102         | Ravi          | 1             |
| 103         | Priya         | 2             |


One Department → Many Employees

### 3. Many-to-Many (M:N)

Many records in Table A can be related to many records in Table B.

Example: Students and Courses.

- One student can enroll in many courses.
- One course can have many students.

We cannot directly create this relationship using only two tables. We use a `junction/bridge` table.

## The SQL JOIN Clause

Joins allow us to retrieve data from multiple tables.

Here are the different types of JOINs in SQL:

- `(INNER) JOIN`: Returns only rows that have matching values in both tables
- `LEFT (OUTER) JOIN`: Returns all rows from the left table, and only the matched rows from the right table
- `RIGHT (OUTER) JOIN`: Returns all rows from the right table, and only the matched rows from the left table
- `FULL (OUTER) JOIN`: Returns all rows when there is a match in either the left or right table
- `CROSS JOIN`: Creates every possible combination of rows.

### CROSS JOIN

A `CROSS JOIN` is different from the other joins.

It does not need a matching column.

It creates every possible combination of rows from both tables.

|          2 | Rahul   |

|       102 | Java        |
|       103 | SQL         |

| name    | course_name |

| Akilesh | Java        |
| Akilesh | SQL         |
| Rahul   | Python      |
| Rahul   | Java        |
| Rahul   | SQL         |


**Important difference**

| JOIN           | What it does                   |
| -------------- | ------------------------------ |
| **INNER JOIN** | Only matching records          |
| **LEFT JOIN**  | All left + matching right      |
| **RIGHT JOIN** | All right + matching left      |
| **FULL JOIN**  | Everything from both tables    |
| **CROSS JOIN** | **Every possible combination** |

## UNION

`UNION` is used to combine the results of two or more SELECT queries into one result.

Unlike JOIN, UNION adds rows vertically.

**Chennai Employees**

| employee_id | name    | department |
| ----------: | ------- | ---------- |
|           1 | Akilesh | IT         |
|           2 | Rahul   | HR         |
|           3 | Priya   | Finance    |

**Bangalore Employees**

| employee_id | name    | department |
| ----------: | ------- | ---------- |
|           4 | Karthik | IT         |
|           5 | Arun    | Sales      |
|           6 | Divya   | HR         |

We want to get all employees from both branches in a single result.

**Result:**
| employee_id | name    | department |
| ----------: | ------- | ---------- |
|           1 | Akilesh | IT         |
|           2 | Rahul   | HR         |
|           3 | Priya   | Finance    |
|           4 | Karthik | IT         |
|           5 | Arun    | Sales      |
|           6 | Divya   | HR         |


**Important Rule**

The two SELECT statements should have:

- Same number of columns
- Columns in the same order
- Corresponding columns should have compatible data types

`UNION` removes duplicates

**UNION vs UNION ALL**

| UNION                               | UNION ALL                     |
| ----------------------------------- | ----------------------------- |
| Combines results                    | Combines results              |
| Removes duplicates                  | Keeps duplicates              |
| May be slower                       | Usually faster                |
| Used when unique results are needed | Used when every row is needed |

### Views

A View is a saved SQL query that behaves like a virtual table.

SELECT student_id, name
FROM students
WHERE course = 'Python';

Views are useful for:

- Simplifying complex queries
- Reusing queries
- Controlling access to data
