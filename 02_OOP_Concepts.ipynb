## 1. Class

### What is a Class?
A **class** is a blueprint/template used to create objects.

### Real-world analogy
think of a building plan.

an architect creates a blueprint that says:

* how many rooms?

* where are the doors?

* where are the windows?

the blueprint itself isn't a building. it is a plan for creating buildings.

similarly, a Python class is a blueprint for **creating objects.**

## concept
A class defines:

* what data an object will have
* what actions an object can perform


### Why use it?
Classes group related **data (attributes)** and **behavior (methods)** together.

### Syntax
```python
class Student:
    pass
```

````python
class student:
    name = "akilesh"
    std = "12th"
student1 = student()
print("Name of the student",student1.name)
print("Class:",student1.std)
````

**Key idea:** The class describes what a Student object can contain or do.


## 2. Object

An **object** is an instance of a class.

## Real-world analogy

The architect's blueprint can be used to construct many actual buildings.

similarly, one class can create many objects.

             Student Class
                   ↓
        ┌──────────┼──────────┐
        ↓          ↓          ↓
     student1   student2   student3
      Aadharsh      vijay      gowtham
### Why use objects?
A class is the blueprint; objects are the actual entities created from that blueprint.


````python
class car: # create a Class
    car_model = "Hyundai"
    car_color = "Green"
    car_milage = "20 km"

    def move_forwar(self):
        print("Moving forward!!")
    def move_backward(self):
        print("Moving Backward!!!")

car1 = car() # Creating Object 1
car2 = car() # Creating Object 2
car3 = car() # Creating Object 3

print(car1.car_model)
print(car2.car_milage)
print(car3.car_color)
````

## 3. Constructor – `__init__()`

`__init__()` is a special method that runs automatically when an object is created.

## Real-world analogy
imagine you're creating a employee ID card.

When a new students joins, you immediately need to enter:

* Name
* Age
* Department
  
The initial information is set when the student is created.
### Why use it?
It initializes an object's starting data.


````python
class car:

    def __init__(self,car_model,car_milage,car_color,car_manufacture_year):
        self.car_model = car_model
        self.car_milage = car_milage
        self.car_color = car_color
        self.car_manufacture_year = car_manufacture_year

    def move_forwar(self):
        print("Moving forward!!")
    def move_backward(self):
        print("Moving Backward!!!")
        

car1 = car("Hyundai",25,"grey",2024) # Creating Object 1
car2 = car("Honda City",22,"green",2025) # Creating Object 2
car3 = car("Thor",20,"black",2020) # Creating Object 3

print("Name of the car 1:",car1.car_model)
print("Milage of the car 1:",car1.car_milage)
print("Name of the car 2:",car2.car_model)
print("Milage of the car 2:",car1.car_milage)
print("Name of the car 3:",car3.car_model)
print("Year car3:",car3.car_manufacture_year)
````

`self` refers to the **current object**.

## 7. Static Methods

A static method does not need `self` or `cls`. It does not depend on an object's data or the class's data.

Use `@staticmethod`.

## Real-world analogy

Suppose a college has a utility function:

Check whether a number is even.

This doesn’t depend on any particular student.

It doesn’t need student information.

That’s a good use case for a static method.

**When to use?**

Use a static method when a function logically belongs to a class but doesn't need access to the object or class.


````python
class Calculator:
    @staticmethod
    def sq(a):
        return a*a

print(Calculator.sq(10))
````

## 8. Class Methods

A class method receives the class itself as `cls`.

Use `@classmethod`.

### When to use it?
Common uses include modifying class-level data and creating alternative constructors.

## Real-world analogy

Imagine a college changes its name.

The change applies to the college as a whole, not just one student.

That’s class-level behavior.


````python
class Student:

    school = "ABC School"

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school


Student.change_school("XYZ School")

print(Student.school)
````

## 9. Properties – `@property`

A property allows a method to be accessed like an attribute.

It is especially useful when you want controlled access to internal data.

## Real-world analogy

Think about a bank account.

You may want to allow someone to see the balance, but you don’t want them to directly modify it however they want.

A property allows controlled access.


````python
class Employee:

    def __init__(self, salary):
        self._salary = salary

    @property
    def salary(self):
        return self._salary


employee = Employee(50000)

print(employee.salary)
````

## 10. Encapsulation

**Encapsulation:** Binding data and methods together into a single unit (class). 
Also called: “Data Hiding”

Python uses naming conventions such as:
- `name` → public
- `_name` → protected-by-convention
- `__name` → name-mangled/private-by-convention
### Real-Life Example 
ATM machine: 
- You can withdraw money 
- But internal bank processing is hidden 

Benefits 
- Security 
- Controlled access 
- Easy maintenance

that's the basic idea of **encapsulation.**

**Key point:** *“Encapsulation protects data from direct unauthorized access.”*


````python
class BankAccount:

    def __init__(self,balance):
        self.__balance = balance

    def deposite(self,amount):
        self.__balance += amount 

    def getbalance(self):
        print("Account Balance:",self.__balance)

acc = BankAccount(1000)
acc.deposite(2000)

acc.getbalance()
````

## 11. Abstraction

**Abstraction** hides implementation details and exposes only the essential interface.

Python provides the `abc` module for formal abstract classes.

## Real-world analogy.
when you drive a car, you press:
    
     START
     
You don't need to understand:
   * Fuel injection
   *    ↓
   * Battery
   *    ↓
   * Starter motor
   *    ↓
   * Engine combustion
   *    ↓
   * Transmission


The complicated implementation is hidden.

Abstraction= show what is needed, hide how it works internally.
     

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car starts with a key or button.")

car = Car()
car.start()
````

### Output
```text
Car starts with a key or button.
```

An abstract method defines **what** a subclass must provide without defining the complete implementation.


## 12. Inheritance

Inheritance allows a child class to reuse and extend functionality from a parent class.

## Real-world analogy

A child can inherit characteristics from parents.

## Why?

To reuse existing code instead of writing it again.

## When?

When there is a genuine "is-a" relationship.

* Dog IS an Animal
* Car IS a Vehicle

### Basic structure
```text
Parent
  ↓
Child
```

````python
class Person:

    def __init__(self,fname,lname):
        self.fname = fname
        self.lname = lname

    def printnames(self):
        print("FirstName:",self.fname,"LastName:",self.lname)

class student(Person):

    def welcome(self):
        print("student Name:",self.fname,self.lname)

student1 = student("Akilesh","E")

student1.welcome()
student1.printnames()
````

## 13. Multiple Inheritance

A class inherits from more than one parent class.

## Real-world analogy

A person might have skills learned from different sources:

* Father → Driving
                         
* Mother → Cooking

* child can access both the skill      

##  Why?

To combine behavior from multiple classes when the design genuinely requires it.

## When?

When one class logically needs functionality from multiple parent classes.

````python
class Father:

    def drive(self):
        print("can drive")

class Mother:

    def cook(self):
        print("Can cook")

class Son(Father,Mother):

    def study(self):
        print("Studying school")

child = Son()

child.cook()
child.drive()
child.study()
````

## 14.Multilevel Inheritance

inheritance occcurs through multiple levels.

## Real-world analogy

Grandparent ---> Parent ---> Child

## Why?

To represent a chain of specialization.

## When?

when each level is meaningful specialization of the previous level.


````python
class Animal:
    def eating(self):
        print("Eating")
class Dog(Animal):
    def bark(self):
        print("Barking")
class puppy(Dog):
    def play(self):
        print("Playing")

dog=puppy()

dog.eating()
dog.bark()
dog.play()

````

## 15.Hierarchical Inheritance

          Animal

 To share common functionality among multiple related classes.

## When?

## 17. Method Overriding

Method overriding happens when a child class provides a different implementation of an inherited method.

## Real-world analogy

**Parent:**
"Animals make sounds."

**Dog:**
"My sound is Woof."

## Why?
To customize inherited behavior.

## When?
When the parent's implementation isn't suitable for a particualr child.

*child class changes parent method behavior* 

This is: 

`Run-time` polymorphism

````python
class Birds:
    def fly(self):
        print("Birds can fly")
class Eagles(Birds):
    def fly(self):
        print("Eagles can fly")

birds=Eagles()

birds.fly()
````

## 19. Composition
Composition means creating a relationship where one object contains another object.

Composition = Has a relationship 
### Real-world analogy

Think about a computer.

A computer is made up of several components:

Computer

 ├── Processor

 ├── RAM
 
 ├── Storage
 
 └── Keyboard

The Computer has these components.

## Why?

To build complex objects by combining smaller objects.

## When?

When one object is naturally made up of other objects.

````python
class Processor:
    def process(self):
        print("Processing the data")

class computer:

    def __init__(self):
        self.processor = Processor()
    def start(self):
        print("Computer Turning ON")
        self.processor.process()

Computer = computer()

Computer.start()
````

## 20.Aggregation

Aggregatiion is also a "has-a"relationship,but the contained objects can exist independently.

Aggregation = One object uses another object, but both objects can exist independently.

## Real-world analogy

A Department has Teachers.

But if the Department is removed, the Teacher can still exist independently.

Department
   
   |
   
   ├── Teacher
   
   ├── Teacher
   
   └── Teacher

The teachers are not completely owned by the department.

## Why?

To represent relationship where objects are grouped together but don't have dependent lifetimes.

## When?

When one objects uses or groups other independently existing objects.

````python
class student:

    def __init__(self,name):
        self.name = name

    def study(self):
        print(self.name,"is studying 12th")

class school:

    def __init__(self,S_name,student):
        self.S_name = S_name
        self.student = student

    def printname(self):
        print("School:",self.S_name)
        self.student.study()

student1 = student("Akilesh")

school1 = school("Sishya matric",student1)

school1.printname()     
        
````

## 23.Super()

Super() gives you access to the parent class implementation from a child class.

## Real-world analogy

Imagine you're a child and your parent already has a particular skill.

instead of rewriting that skill,you say:

"I'll use my parent's implementation."

## Why?
To reuse parent functionality.

## When?
Especially when extending a parent's constructor or method.

````python
class Person:

    def __init__(self,fname,lname):
        self.fname=fname
        self.lname=lname
    def printname(self):
        print(self.fname,self.lname)

class student(Person):

    def __init__(self,fname,lname,year):
        super().__init__(fname,lname)
        self.year = year

    def printstudent(self):
        print("Name :",self.fname,self.lname,"year:",self.year)

x = student("Akilesh","E",2024)

x.printname()
x.printstudent()
        
````

## 26. `__len__()`

`__len__()` defines what should happen when python's len() function is called on our object.

## Real-world analogy

suppose we create a team object.

len(team)

to return the number of team members.

## Why?

To make custom objects work naturally with Python's len() function.

## When?

## 28.Singleton pattern

The Singleton Pattern ensures that a class has only one instance.

## Real-world analogy

Think of a school principal.

A school should have only one principal at a time.

Even if different teachers ask for the principal:

Teacher 1 ──┐

Teacher 2   ---> Same Principal

Teacher 3 ──┘

Similarly, in the Singleton Pattern:

object1 ──┐

object2 ──┼──> Same Object

object3 ──┘

## Why?

To provide a single shared instance where having multiple instances would be unnecessary or undesirable.

## When?

potential examples include:
* configuration management
* certain resource managers
* some application - wide services

however, singleton should be used carefully
because it introduces shared/global state.


Singleton = Single + Instance

- Single → Only one
- Instance → Object

````python
class Principle:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            print("Creating Principle Object")
            cls._instance = super().__new__(cls)
        return cls._instance

principle1 = Principle()

principle2 = Principle()

principle3 = Principle()

print (principle1 is principle2)
````
